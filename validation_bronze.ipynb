```python
%sql
WITH src AS (
  SELECT 'customer' AS table_name, COUNT(*) AS row_count FROM samples.tpch.customer
  UNION ALL SELECT 'orders'  , COUNT(*) FROM samples.tpch.orders
  UNION ALL SELECT 'lineitem', COUNT(*) FROM samples.tpch.lineitem
  UNION ALL SELECT 'part'    , COUNT(*) FROM samples.tpch.part
  UNION ALL SELECT 'partsupp', COUNT(*) FROM samples.tpch.partsupp
  UNION ALL SELECT 'supplier', COUNT(*) FROM samples.tpch.supplier
  UNION ALL SELECT 'nation'  , COUNT(*) FROM samples.tpch.nation
  UNION ALL SELECT 'region'  , COUNT(*) FROM samples.tpch.region
),
brz AS (
  SELECT 'customer' AS table_name, COUNT(*) AS row_count FROM assignment1.bronze.customer
  UNION ALL SELECT 'orders'  , COUNT(*) FROM assignment1.bronze.orders
  UNION ALL SELECT 'lineitem', COUNT(*) FROM assignment1.bronze.lineitem
  UNION ALL SELECT 'part'    , COUNT(*) FROM assignment1.bronze.part
  UNION ALL SELECT 'partsupp', COUNT(*) FROM assignment1.bronze.partsupp
  UNION ALL SELECT 'supplier', COUNT(*) FROM assignment1.bronze.supplier
  UNION ALL SELECT 'nation'  , COUNT(*) FROM assignment1.bronze.nation
  UNION ALL SELECT 'region'  , COUNT(*) FROM assignment1.bronze.region
)
SELECT s.table_name,
       s.row_count AS source_cnt,
       b.row_count AS bronze_cnt,
       CASE WHEN s.row_count = b.row_count THEN 'PASS' ELSE 'FAIL' END AS status
FROM src s JOIN brz b USING (table_name)
ORDER BY s.table_name;
```
