```python
%sql
SELECT SUM(li.quantity) AS total_items
FROM orders o
JOIN lineitem li ON li.order_key = o.order_key
WHERE o.customer_key = 446044;
```

```python
%sql
SELECT MIN(orderdate) AS first_purchase_date
FROM orders
WHERE customer_key = 564787;
```

```python
%sql
SELECT COUNT(*) AS total_orders
FROM orders
WHERE customer_key = 85909;
```

```python
%sql
WITH order_counts AS (
  SELECT o.customer_key, COUNT(*) AS orders_cnt
  FROM orders o
  GROUP BY o.customer_key
),
top_c AS (
  SELECT customer_key, orders_cnt
  FROM order_counts
  ORDER BY orders_cnt DESC, customer_key ASC
  LIMIT 1
)
SELECT
  tc.customer_key,
  tc.orders_cnt,
  n.name AS nationality
FROM top_c tc
JOIN customer c ON c.customer_key = tc.customer_key
JOIN nation   n ON n.nation_key   = c.nation_key;
```
